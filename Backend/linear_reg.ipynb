## Linear Regression

```python
from langchain_classic.vectorstores import sklearn
from sklearn.datasets import fetch_california_housing
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = fetch_california_housing()
```

```python
type(df)
```

```python
df
```

```python
datasets = pd.DataFrame(df.data)
datasets.columns = df.feature_names
datasets.head()
```

```python
datasets['price'] = df.target
```

```python
datasets.head()
```

```python
#dividing the dataset into independent and dependent features
X = datasets.iloc[:, :-1]
Y = datasets.iloc[:,-1]
```

```python
X.head()
```

```python
Y.head()
```

```python
##Linear Regression
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
linear_reg = LinearRegression()
mse = cross_val_score(linear_reg, X, Y, scoring='neg_mean_squared_error', cv=5)
# print(mse)

mean_mse = np.mean(mse)
print(mean_mse)
```

Ridge Regression

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV

```

```python
ridge = Ridge()
params = {'alpha': [1e-15, 1e-10, 1e-8, 1e-3, 1e-2, 1, 5, 10, 20,30,35,40,50,70,80,85,100]}
ridge_reg = GridSearchCV(ridge, params, scoring='neg_mean_squared_error', cv=10)
ridge_reg.fit(X, Y)
```

```python
print(ridge_reg.best_params_)
print(ridge_reg.best_score_)
```

```python
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV

lasso = Lasso()
params = {'alpha': [1e-15, 1e-10, 1e-8, 1e-3, 1e-2, 1, 5, 10, 20,30,35,40,50,70,80,85,100]}
lasso_reg = GridSearchCV(lasso, params, scoring='neg_mean_squared_error', cv=10)
lasso_reg.fit(X, Y)
```

```python
print(lasso_reg.best_params_)
print(lasso_reg.best_score_)
```
